# Building a First Supervised Machine Learning Model



```python
import pandas as pd, numpy as np

url = "hf://datasets/samyuktha01/Indian_Railway_maintance/indian_railway_predictive_maintenance_100k.csv"
df = pd.read_csv(url)
df.head()
```

```python
# Features and Target
features = ["inspection_score", "sensor_health_index"]
target = "risk_score"

# The problem is a regression problem because the target variable, risk_score,
# is predicting a numerical value rather than a category.
```

```python
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

X = df[features]
y = df[target]

# Splitting the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)

# Training supervised learning models
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

random_forest_model = RandomForestRegressor(n_estimators=100, random_state=42)
random_forest_model.fit(X_train, y_train)

# Making predictions
pred_linear = linear_model.predict(X_test)
pred_rf = random_forest_model.predict(X_test)

results = pd.DataFrame({
    "Actual Risk Score": y_test,
    "Linear Prediction": pred_linear,
    "Random Prediction": pred_rf})
print(results.head(5))
```

```python
# Evaluation of the models
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mean_absolute_error(y_test, pred_linear),
            mean_absolute_error(y_test, pred_rf)],

    "RMSE": [np.sqrt(mean_squared_error(y_test, pred_linear)),
             np.sqrt(mean_squared_error(y_test, pred_rf))],

    "R2 Score": [r2_score(y_test, pred_linear),
                 r2_score(y_test, pred_rf)]
    })

print("Comparison of the perfomance of the modules")
print(results.round(3))
```

## **Discussion of the results**

### Linear regression performed better on all three metrics.

### Linear Regression outperformed Random Forest on MAE, RMSE and R², using inspection_score and sensor_health_index to predict risk_score.


* Its Mean Absolute Error is about 0.31 lower, indicating its predictions were closer to the actual risk scores on average.


*   Its Root Mean Squared Error is about 0.40 lower, indicating smaller prediction errors overall.
*   Its R² score of 0.681 explains 68.1% of the variation in risk scores, against 61.95 for the Random Forest.

### These results suggest that Linear Regression captures the relationships between the selected features (*inspection_score* and *sensor_health_index*), and the target variable (*risk_score*) more effectively in this dataset.

### **Conclusion**
 #### Linear Regression is the preferred model for this problem, based on the three evaluation metrics.
