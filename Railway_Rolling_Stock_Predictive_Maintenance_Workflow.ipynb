# Project Problem
Predictive maintenance of railway rolling stock (wagons, locomotives and coaches)

### Proposed Machine Learning Workflow
    A[Rolling Stock Data Collection] --> B[Data Importation]
    B --> C[Data Cleaning & Preparation]
    C --> D[Feature Selection]
    D --> E[Identify Target Variable]
    E --> F[Split Data into Training & Testing Sets]
    F --> G[Train Machine Learning Model]
    G --> H[Model Evaluation]
    H --> I{Model Performance Acceptable?}
    I -- No --> D
    I -- Yes --> J[Predict Maintenance Requirement]
    J --> K[Maintenance Planning & Decision Support]
    K --> L[Improved Reliability, Availability & Maintenance]

#### Workflow Analysis
The proposed workflow shows how rolling stock data will be processed and used to develop a predictive maintenance model.


1. **Rolling Stock Data Collection**
- Data is collected from rolling stock inspection records, maintenance history, operational data, and sensor measurements.

2. **Data Importation**
- The available dataset is imported into the Python/Colab environment for analysis.

3. **Data Cleaning & Preparation**
- The data is checked for missing values, incorrect entries, duplicates, and inconsistency. The data is then prepared for machine-learning analysis.

4. **Feature Selection**
- Relevant variables such as wheel wear, bearing temperature, brake condition, equipment age, inspection results, and time since last maintenance are selected as potential inputs to the model.

5. **Identify Target Variable**
- The target is set to whether the rolling stock requires maintenance or not.
`0` may represent maintenance not required and `1` may represent maintenance required.

6. **Training and Testing**
- The dataset is divided into training and testing data. The training data is used to teach the model, while the testing data is used to evaluate how well it performs on previously unseen data.

7. **Model Training**
- A suitable ML classification algorithm is trained to learn relationships between the rolling stock condition data and maintenance requirements.

8. **Model Evaluation**
- The model's predictions are evaluated using appropriate performance measures to determine whether it can reliably identify rolling stock requiring maintenance.

9. **Maintenance Prediction**
- Once the model performs satisfactorily, it can be used to predict whether a wagon, locomotive, or coach is likely to require maintenance based on its current condition and operational data.

10. **Maintenance Decision Support**
- The predictions can support maintenance planning by helping identify equipment that may require inspection or intervention before a significant failure occurs.

11. **Expected Outcome**
- The overall objective is to improve rolling stock reliability, availability, and maintenance effectiveness, while reducing unexpected failures and unplanned downtime.
